```python

import pandas as pd
from pathlib import Path


class Dataset:
    """
    Handles loading and accessing datasets for the data cleaning library.
    """

    def __init__(self, data=None, filepath=None):
        self._data = data if data is not None else pd.DataFrame()
        self._filepath = filepath

    @classmethod
    def load_csv(cls, filepath):
        """
        Loads a CSV file and returns a Dataset object.
        """
        filepath = Path(filepath)

        if not filepath.exists():
            raise FileNotFoundError(f"File not found: {filepath}")

        if not filepath.is_file():
            raise ValueError("The provided path is not a file.")

        try:
            data = pd.read_csv(filepath)

            if data.empty:
                raise ValueError("The CSV file is empty.")

            return cls(data, str(filepath))

        except pd.errors.EmptyDataError:
            raise ValueError("The CSV file contains no data.")

        except pd.errors.ParserError:
            raise ValueError("Unable to read the CSV file. Check its format.")

    def preview(self, n=5):
        """Displays the first n rows of the dataset."""
        return self._data.head(n)

    def get_columns(self):
        """Returns the column names."""
        return self._data.columns.tolist()

    def get_shape(self):
        """Returns the number of rows and columns."""
        return self._data.shape

    def get_info(self):
        """Returns information about the dataset."""
        return self._data.info()

    def get_data(self):
        """Returns the actual DataFrame."""
        return self._data

    def __len__(self):
        """Returns the number of rows."""
        return len(self._data)

    def __repr__(self):
        """Returns a readable representation of the dataset."""
        return f"Dataset(rows={len(self)}, columns={len(self._data.columns)})"

    def __getitem__(self, key):
        """Allows access to dataset columns."""
        return self._data[key]
```
